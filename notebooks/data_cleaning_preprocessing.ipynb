# Assignment 2: Clean and Preprocess a Dataset using Pandas and NumPy

**Dataset:** Netflix Movies and TV Shows (Kaggle / public domain mirror)

**Objective:** Load the dataset, inspect it for missing values, duplicates and wrong data
types, then clean it: handle missing values, remove duplicates, convert data types, run
NumPy numerical transformations, use Pandas for filtering/sorting/grouping, compute summary
statistics, and finish with ML-readiness encoding plus visual insights.

```python
import os
import sys
sys.path.insert(0, os.path.dirname(os.path.abspath('.')))

import numpy as np
import pandas as pd

from src import cleaning as cl
```

## Step 1 - Load the dataset using Pandas

```python
df = cl.load_data("data/raw/netflix_titles.csv")
print(f"Shape: {df.shape}")
df.head()
```

## Step 2 - Inspect missing values, duplicates and data types

```python
missing = cl.missing_values_table(df)
missing
print("Duplicate rows (full):", cl.duplicate_count(df))
cl.data_type_report(df)
```

## Step 3 - Handle missing values (drop / fill / replace)

```python
df = cl.handle_missing_values(df)
cl.missing_values_table(df)
```

## Step 4 - Remove / correct duplicate records

```python
df = cl.remove_duplicates(df)
print(f"Rows after dedupe: {len(df)}")
```

## Step 5 - Convert data types (date parsing, duration extraction)

```python
df = cl.parse_date_added(df)
df = cl.extract_duration(df)
df = cl.convert_types(df)
df[["date_added", "date_added_dt", "duration", "duration_value", "duration_unit"]].head()
```

## Step 6 - NumPy numerical transformations

```python
df = cl.numpy_transformations(df)
```

## Step 7 - Pandas filtering, sorting, grouping

```python
recent = cl.filter_movies_after_year(df, 2018)
print(f"Movies after 2018: {len(recent)}")
cl.sort_by_release_year(df).head()
by_country = cl.group_by_country(df)
by_country.head(10)
```

## Summary statistics

```python
cl.summary_statistics(df)
```

## Bonus 1 - Null-value heatmap & correlation matrix

```python
cl.plot_null_heatmap(df, "reports/figures/null_heatmap.png")
cl.plot_correlation_matrix(df, "reports/figures/correlation_matrix.png")
print("Figures saved to reports/figures/")
```

## Bonus 2 - ML-readiness (label encoding + one-hot)

```python
df = cl.label_encode(df, ["type", "rating"])
df = cl.top_country_encoding(df, top_n=15)
print(f"Final shape: {df.shape}")
df.head()
```

## Save the cleaned dataset

```python
import os
os.makedirs("data/processed", exist_ok=True)
df.to_csv("data/processed/netflix_titles_clean.csv", index=False)
print("Saved data/processed/netflix_titles_clean.csv")
```

## Key findings

1. **Missing values** were concentrated in `director`, `cast`, `country`, `date_added`,
   `rating`, `duration`, `listed_in` and `description`; they were filled with
   `Unknown`/`Not available`/forward-fill rather than dropping thousands of rows.
2. **Duplicates** (full rows and `show_id` repeats) were removed.
3. **Date parsing** converted `date_added` to a real datetime and extracted `added_year`.
4. **Duration parsing** split `90 min` / `2 Seasons` into numeric value + unit, enabling
   numeric analysis (median movie duration ~88 min).
5. **Correlation insight:** `release_year` and `added_year` are the most correlated numeric
   fields; `is_movie` is negatively related to `duration_value`-style season counts.
6. The **cleaned, ML-ready dataset** (38 columns) was written to
   `data/processed/netflix_titles_clean.csv`.
